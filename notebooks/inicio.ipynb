```python
import ipywidgets as widgets
from IPython.display import display

boton = widgets.Button(description="Sumar")
salida = widgets.Output()

contador = 0

def sumar(boton):
    global contador

    contador += 1

    with salida:
        salida.clear_output()
        print(f"Puntos: {contador}")

boton.on_click(sumar)

display(boton, salida)
```

```python
import ipywidgets as widgets
from IPython.display import display, HTML


# -------------------------
# Estado del jugador
# -------------------------

x = 1
y = 1


# -------------------------
# Configuración visual
# -------------------------

TAMANO_CASILLA = 100
RADIO_JUGADOR = 25


# -------------------------
# Dibujar el tablero
# -------------------------

def dibujar():
    svg = f"""
    <svg width="300" height="300"
         viewBox="0 0 300 300"
         xmlns="http://www.w3.org/2000/svg">

        <!-- Fondo -->
        <rect
            x="0"
            y="0"
            width="300"
            height="300"
            fill="white"
        />

        <!-- Líneas verticales -->
        <line
            x1="100" y1="0"
            x2="100" y2="300"
            stroke="black"
            stroke-width="2"
        />

        <line
            x1="200" y1="0"
            x2="200" y2="300"
            stroke="black"
            stroke-width="2"
        />

        <!-- Líneas horizontales -->
        <line
            x1="0" y1="100"
            x2="300" y2="100"
            stroke="black"
            stroke-width="2"
        />

        <line
            x1="0" y1="200"
            x2="300" y2="200"
            stroke="black"
            stroke-width="2"
        />

        <!-- Jugador -->
        <circle
            cx="{x * TAMANO_CASILLA + TAMANO_CASILLA / 2}"
            cy="{(2 - y) * TAMANO_CASILLA + TAMANO_CASILLA / 2}"
            r="{RADIO_JUGADOR}"
            fill="red"
        />

    </svg>
    """

    return HTML(svg)


# -------------------------
# Crear botones
# -------------------------

def crear_boton(texto):
    return widgets.Button(
        description=texto,
        layout=widgets.Layout(
            width="60px",
            height="50px"
        ),
        style=widgets.ButtonStyle(
            button_color="#d0d0d0"
        )
    )


arriba = crear_boton("↑")
abajo = crear_boton("↓")
izquierda = crear_boton("←")
derecha = crear_boton("→")


# -------------------------
# Pantalla
# -------------------------

pantalla = widgets.Output()


# -------------------------
# Actualizar pantalla
# -------------------------

def actualizar():
    pantalla.clear_output(wait=True)

    with pantalla:
        display(dibujar())


# -------------------------
# Movimiento
# -------------------------

def mover_arriba(boton):
    global y

    if y < 2:
        y += 1

    actualizar()


def mover_abajo(boton):
    global y

    if y > 0:
        y -= 1

    actualizar()


def mover_izquierda(boton):
    global x

    if x > 0:
        x -= 1

    actualizar()


def mover_derecha(boton):
    global x

    if x < 2:
        x += 1

    actualizar()


# -------------------------
# Conectar botones
# -------------------------

arriba.on_click(mover_arriba)
abajo.on_click(mover_abajo)
izquierda.on_click(mover_izquierda)
derecha.on_click(mover_derecha)


# -------------------------
# Crear cruceta
# -------------------------

controles = widgets.GridspecLayout(
    2,
    3,
    width="190px"
)

controles[0, 1] = arriba

controles[1, 0] = izquierda
controles[1, 1] = abajo
controles[1, 2] = derecha


# -------------------------
# Arrancar
# -------------------------

actualizar()

display(pantalla)
display(controles)
```
